# Imports y Configuración de Rutas

```python
import os
import numpy as np
import tensorflow as tf
import pickle
import chess.pgn
import pandas as pd
import matplotlib.pyplot as plt

# --- CONFIGURACIÓN DE RUTAS ---
CHECKPOINTS_DIR = "../trained_models/"
TOKENIZER_PATH = "../trained_models/tokenizer.pkl"
TEST_DATA_DIR = "../test_data/"
MAX_SEQUENCE_LENGTH = 50
# -----------------------------

# Buscar automáticamente todos los archivos de checkpoints en la carpeta
checkpoint_files = sorted([f for f in os.listdir(CHECKPOINTS_DIR) if f.endswith('.keras')])

print(f"Se encontraron {len(checkpoint_files)} checkpoints para evaluar:")
for file in checkpoint_files:
    print(f" - {file}")
```

# Cargar el Tokenizer y las Partidas de Prueba

```python
# Cargar el tokenizador
with open(TOKENIZER_PATH, 'rb') as f:
    tokenizer = pickle.load(f)
print(f"✅ Tokenizador cargado. Vocabulario: {len(tokenizer.word_index) + 1} palabras.")

# Cargar todas las partidas de prueba de la carpeta test_data
partidas_test = []
print(f"\nCargando partidas de prueba desde: {TEST_DATA_DIR}")
for pgn_filename in os.listdir(TEST_DATA_DIR):
    if pgn_filename.endswith(".pgn"):
        pgn_path = os.path.join(TEST_DATA_DIR, pgn_filename)
        with open(pgn_path) as pgn_file:
            try:
                partida = chess.pgn.read_game(pgn_file)
                if partida:
                    movimientos_reales = []
                    board = partida.board()
                    for move in partida.mainline_moves():
                        movimientos_reales.append(board.san(move))
                        board.push(move)
                    partidas_test.append(movimientos_reales)
            except Exception as e:
                print(f"Error al leer {pgn_filename}: {e}")

print(f"✅ Se cargaron {len(partidas_test)} partidas para la evaluación.")
```

# El "Motor" de Evaluación

```python
# --- Función de Predicción ---
# La traemos de nuestro notebook anterior para usarla aquí
def predecir_siguientes_movimientos(model, tokenizer, sequence_text, top_k=5):
    sequence_num = tokenizer.texts_to_sequences([sequence_text])
    sequence_padded = tf.keras.preprocessing.sequence.pad_sequences(
        sequence_num, maxlen=MAX_SEQUENCE_LENGTH, padding='post'
    )
    prediction = model.predict(sequence_padded, verbose=0)[0]
    top_indices = np.argsort(prediction)[-top_k:][::-1]
    predicted_moves = [tokenizer.index_word.get(i, "[DESCONOCIDO]") for i in top_indices]
    return predicted_moves

# --- Función de Evaluación ---
# Esta función toma un modelo y un set de partidas, y calcula las métricas
def evaluar_modelo(model, tokenizer, partidas_test):
    top1_hits = 0
    top5_hits = 0
    total_moves = 0

    for partida in partidas_test:
        # Evaluamos a partir de la jugada 5 para tener algo de contexto
        for i in range(5, len(partida)):
            historial = partida[:i]
            jugada_correcta = partida[i]
            total_moves += 1
            
            # Obtenemos las 5 mejores predicciones del modelo
            predicciones = predecir_siguientes_movimientos(model, tokenizer, historial, top_k=5)
            
            # Contamos los aciertos
            if jugada_correcta == predicciones[0]:
                top1_hits += 1
            if jugada_correcta in predicciones:
                top5_hits += 1
    
    # Calculamos la precisión final
    acc_top1 = top1_hits / total_moves if total_moves > 0 else 0
    acc_top5 = top5_hits / total_moves if total_moves > 0 else 0
    
    return acc_top1, acc_top5

# --- Bucle Principal de Evaluación ---
# Iteramos sobre cada archivo de checkpoint que encontramos en la Celda 1

resultados_evaluacion = []

for checkpoint_file in checkpoint_files:
    # Extraemos el número de la época del nombre del archivo
    epoch_num = int(os.path.basename(checkpoint_file).split('_')[2].split('.')[0])
    print(f"\n--- Evaluando Checkpoint: {checkpoint_file} (Época {epoch_num}) ---")
    
    # Cargamos el modelo de esta época
    model_path = os.path.join(CHECKPOINTS_DIR, checkpoint_file)
    model = tf.keras.models.load_model(model_path)
    
    # Ejecutamos la evaluación
    acc_top1, acc_top5 = evaluar_modelo(model, tokenizer, partidas_test)
    
    # Guardamos los resultados
    resultados_evaluacion.append({
        "epoch": epoch_num,
        "top1_accuracy": acc_top1,
        "top5_accuracy": acc_top5
    })
    
    print(f"Resultados Época {epoch_num}: Top-1 Acc = {acc_top1:.2%}, Top-5 Acc = {acc_top5:.2%}")

print("\n✅ Evaluación de todos los checkpoints completada.")
```

# Visualicemos la Historia

```python
# Convertimos los resultados a un DataFrame de Pandas para facilitar el manejo
df_resultados = pd.DataFrame(resultados_evaluacion).sort_values(by='epoch')

# --- Creación del Gráfico ---
plt.style.use('seaborn-v0_8-whitegrid')
fig, ax = plt.subplots(figsize=(14, 8))

# Graficar la precisión Top-1 y Top-5
ax.plot(df_resultados['epoch'], df_resultados['top1_accuracy'], label='Precisión Top-1', marker='o', linestyle='--')
ax.plot(df_resultados['epoch'], df_resultados['top5_accuracy'], label='Precisión Top-5', marker='o', linestyle='-')

# Encontrar y marcar el mejor punto
best_epoch = df_resultados.loc[df_resultados['top5_accuracy'].idxmax()]
ax.axvline(x=best_epoch['epoch'], color='r', linestyle='--', label=f"Mejor Época ({int(best_epoch['epoch'])}) en Top-5 Acc")
ax.scatter(best_epoch['epoch'], best_epoch['top5_accuracy'], color='red', s=150, zorder=5)

# Títulos y etiquetas
ax.set_title('Rendimiento del Modelo por Época en el Set de Validación', fontsize=18)
ax.set_xlabel('Época de Entrenamiento', fontsize=14)
ax.set_ylabel('Precisión (Accuracy)', fontsize=14)
ax.set_xticks(df_resultados['epoch']) 
ax.legend(fontsize=12)
ax.grid(True)

plt.show()

# Imprimir el mejor resultado
print("\n--- Mejor Modelo Encontrado ---")
print(f"El mejor rendimiento se alcanzó en la Época: {int(best_epoch['epoch'])}")
print(f"Precisión Top-1 en esa época: {best_epoch['top1_accuracy']:.2%}")
print(f"Precisión Top-5 en esa época: {best_epoch['top5_accuracy']:.2%}")
```

# De los Números a la "Inteligencia": Análisis Cualitativo

```python
# --- ANÁLISIS CUALITATIVO (VERSIÓN CON LÓGICA DE HISTORIAL CORREGIDA) ---
import chess.pgn
import io
import tensorflow as tf
import numpy as np
import os

print("\n\n--- Iniciando Análisis Cualitativo Definitivo ---")

# --- 1. DATOS DE LA PARTIDA ---
pgn_texto = """
[Event "Third Rosenwald Trophy"]
[Site "New York, NY USA"]
[Date "1956.10.17"]
[Round "8"]
[White "Donald Byrne"]
[Black "Robert James Fischer"]
[Result "0-1"]
[ECO "D92"]

1. Nf3 Nf6 2. c4 g6 3. Nc3 Bg7 4. d4 O-O 5. Bf4 d5 6. Qb3 dxc4 7. Qxc4 c6 8. e4
Nbd7 9. Rd1 Nb6 10. Qc5 Bg4 11. Bg5 Na4 12. Qa3 Nxc3 13. bxc3 Nxe4 14. Bxe7 Qb6
15. Bc4 Nxc3 16. Bc5 Rfe8+ 17. Kf1 Be6
"""
partida = chess.pgn.read_game(io.StringIO(pgn_texto))
board = partida.board()

# #######################################################################
# ### INICIO DEL CAMBIO ###
# #######################################################################
# --- 2. PROCESAMIENTO AUTOMÁTICO DEL PGN (CORREGIDO) ---
mainline_moves = list(partida.mainline_moves())
historial_partida = []
# Hacemos las primeras 33 medias jugadas (hasta 17. Kf1 inclusive)
for i in range(33):
    move = mainline_moves[i]
    # Obtenemos la notación en el contexto actual, ANTES de hacer la jugada
    historial_partida.append(board.san(move))
    # Hacemos la jugada para avanzar el tablero a la siguiente posición
    board.push(move)
# #######################################################################
# ### FIN DEL CAMBIO ###
# #######################################################################

# Ahora 'board' contiene la posición exacta después de 17. Kf1
fen_posicion = board.fen()
jugada_brillante = "Be6"

# --- 3. ANÁLISIS DEL MODELO ---
modelos_a_comparar = {
    "Mejor Modelo (Época 7)": "../trained_models/modelo_epoch_07.keras",
    "Modelo Sobreajustado (Época 11)": "../trained_models/modelo_epoch_11.keras"
}
tokenizer_path = "../trained_models/tokenizer.pkl"
with open(tokenizer_path, 'rb') as f:
    tokenizer = pickle.load(f)

# (Función de predicción que ya teníamos)
def predecir_siguientes_movimientos(model, tokenizer, sequence_text, top_k=5):
    sequence_num = tokenizer.texts_to_sequences([sequence_text])
    sequence_padded = tf.keras.preprocessing.sequence.pad_sequences(
        sequence_num, maxlen=50, padding='post'
    )
    prediction = model.predict(sequence_padded, verbose=0)[0]
    top_indices = np.argsort(prediction)[-top_k:][::-1]
    predicted_moves = [tokenizer.index_word.get(i, "[DESCONOCIDO]") for i in top_indices]
    return predicted_moves

print(f"\nAnalizando la posición (Turno de Negras):")
print(board)
print(f"\nFEN extraído automáticamente: {fen_posicion}")
print(f"La jugada genial de Fischer en esta posición fue: !! {jugada_brillante} !!")

for nombre, ruta_modelo in modelos_a_comparar.items():
    print(f"\n--- Sugerencias de: {nombre} ---")
    modelo_cargado = tf.keras.models.load_model(ruta_modelo)
    sugerencias = predecir_siguientes_movimientos(modelo_cargado, tokenizer, historial_partida, top_k=5)
    for i, jugada in enumerate(sugerencias, 1):
        print(f"  {i}. {jugada}")
```

## Conclusiones 

### El Mejor Modelo (Época 7): Piensa como un Jugador Fuerte

Este modelo se enfrenta a la posición crítica de Fischer y, aunque no encuentra la jugada inmortal Be6, sus ideas son excelentes, tácticas y todas legales.

Bxf3: Su primera sugerencia es una jugada muy fuerte. Elimina un defensor clave del rey blanco (el caballo de f3) y daña permanentemente la estructura de peones de White. Es una idea agresiva y coherente.

e6, Nxa2, Nd5, Bf5: El resto de sus sugerencias son todas jugadas lógicas, activas y buenas que cualquier jugador humano consideraría seriamente para mejorar la posición.

Opinión: Este modelo está "entendiendo" la posición a un nivel muy alto. Sus ideas tienen un plan detrás. Juega un ajedrez real, coherente y peligroso.

### El Modelo Sobreajustado (Época 11): La Mente Colapsada

Aquí vemos la prueba definitiva de por qué más entrenamiento fue perjudicial. Este modelo ignora completamente el tablero.

Ninguna de sus 5 mejores sugerencias es una jugada legal o sensata para las Negras en esta posición.

O-O (Enroque): Es ilegal, las negras ya enrocaron en la jugada 4.

Nf6, d5, Nc6: Son jugadas de apertura que son imposibles de hacer desde la posición actual.

Nf3: ¡Es una jugada de las Blancas!

Opinión: El modelo está completamente "alucinando". Ha perdido toda capacidad de entender el contexto y solo está recitando las jugadas de apertura más comunes que memorizó durante el entrenamiento. 
    Es la demostración visual perfecta del sobreajuste.
